# Low-Light Image Enhancement

Load the local Zero-DCE model, train it on images in `data/raw/`, and preview an enhanced result.

```python
from pathlib import Path
import sys
import torch

ROOT = Path.cwd()
if not (ROOT / 'src').is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))

from low_light_enhancement.data import LowLightDataset
from low_light_enhancement.losses import ZeroDCELoss
from low_light_enhancement.models import ZeroDCE
```

```python
dataset = LowLightDataset(ROOT / 'data/raw')  # Add low-light images first
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = ZeroDCE().to(device)
criterion = ZeroDCELoss().to(device)
print(f'{len(dataset)} training images on {device}')
```
